```python
h5_path = "./Code/other_models/TranSiGen/data/Meisheng_used_data/processed_data.h5"

with h5py.File(h5_path, "r") as f:
    smiles_raw = np.asarray(f["canonical_smiles"])        # dtype='S…'

# convert bytes → str (if not already)
if isinstance(smiles_raw[0], (bytes, np.bytes_)):
    smiles = smiles_raw.astype(str)
else:
    smiles = smiles_raw                                    # already str

print("Total rows read:", len(smiles))
print("Example:", smiles[:3])
```

```python
# dict.fromkeys keeps first occurrence order in Python 3.7+
unique_smiles = list(dict.fromkeys(smiles))
print("Unique SMILES:", len(unique_smiles))
```

```python
idx2smi = {i: s for i, s in enumerate(unique_smiles)}

# quick peek
print("First 5 items:")
list(idx2smi.items())[:5]
```

```python
pickle_path = "./Code/other_models/TranSiGen/data/Meisheng_used_data/idx2smi.pickle"

with open(pickle_path, "wb") as fh:
    pickle.dump(idx2smi, fh, protocol=pickle.HIGHEST_PROTOCOL)

print("Saved to", pickle_path)
```

```python
with open(pickle_path, "rb") as fh:
    idx2smi_check = pickle.load(fh)

print("Length matches:", len(idx2smi_check) == len(idx2smi))
print(type(idx2smi_check))
print("\nFirst 5 again:")
list(idx2smi_check.items())[:5]
```
